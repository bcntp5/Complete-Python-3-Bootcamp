```python
#1. Generate Right skewed data
training_hours_completed = np.random.exponential(scale=10, size=1000)

#2. apply log1p transformation to normalize the distrubution
transformed_data = np.log1p(training_hours_completed)

#3. Plot the before and after histograms
fig, axes = plt.subplots(1, 2, figsize=(12, 5))

# Plot the original skewed data on the first subplot (axes[0])
axes[0].hist(training_hours_completed, bins=30, color='blue', edgecolor='black')
axes[0].set_title('Original Right-Skewed Data')
axes[0].set_xlabel('Training Hours Completed')
axes[0].set_ylabel('Frequency')

# Loglp1p Transformed Historgram

axes[1].hist(transformed_data, bins=30, color= 'green', edgecolor= 'black')
axes[1].set_title('Log1p Transformed Data')
axes[1].set_xlabel('Log1p(Training Hours Completed)')
axes[1].set_ylabel('Frequency')

plt.tight_layout()
plt.show()

```

As the numbers get bigger the logramic scale slower due to the exponents. Applying the lp pulls the outside numbers back toward the center. For near 0 the transformation stretches the spacing. This is away to keep the data clean and the numbers close to zero with out messing up he plotted numbers.
